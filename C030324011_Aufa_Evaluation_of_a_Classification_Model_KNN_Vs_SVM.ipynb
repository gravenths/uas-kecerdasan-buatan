```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
```

```python
df_pelanggan = pd.read_csv('UAS_Churn_Modelling.csv')
```

```python
display(df_pelanggan.head())
```

```python
df_pelanggan.info()
```

```python
df_pelanggan = df_pelanggan.dropna().reset_index(drop=True)
```

```python
pengkode_teks = LabelEncoder()

kolom_kategori = df_pelanggan.select_dtypes(include=['object']).columns

for k in kolom_kategori:
    df_pelanggan[k] = pengkode_teks.fit_transform(df_pelanggan[k])
```

```python
fitur_X = df_pelanggan.iloc[:, :-1]
label_y = df_pelanggan.iloc[:, -1]
```

```python
X_train_data, X_test_data, y_train_data, y_test_data = train_test_split(
    fitur_X,
    label_y,
    test_size=0.2,
    random_state=42
)
```

```python
standarisasi = StandardScaler()

X_train_data = standarisasi.fit_transform(X_train_data)
X_test_data = standarisasi.transform(X_test_data)
```

```python
# KNN
algoritma_knn = KNeighborsClassifier(n_neighbors=5)
algoritma_knn.fit(X_train_data, y_train_data)

prediksi_knn = algoritma_knn.predict(X_test_data)
```

```python
skor_knn = accuracy_score(y_test_data, prediksi_knn)

print("=== HASIL EVALUASI KNN ===")
print(f"Tingkat Akurasi: {skor_knn:.4f}\n")

print("Laporan Klasifikasi:")
print(classification_report(y_test_data, prediksi_knn))

print("Confusion Matrix:")
print(confusion_matrix(y_test_data, prediksi_knn))
```

```python
# SVM
algoritma_svm = SVC(kernel='rbf', random_state=42)
algoritma_svm.fit(X_train_data, y_train_data)

prediksi_svm = algoritma_svm.predict(X_test_data)
```

```python
skor_svm = accuracy_score(y_test_data, prediksi_svm)

print("=== HASIL EVALUASI SVM ===")
print(f"Tingkat Akurasi: {skor_svm:.4f}\n")

print("Laporan Klasifikasi:")
print(classification_report(y_test_data, prediksi_svm))

print("Confusion Matrix:")
print(confusion_matrix(y_test_data, prediksi_svm))
```

```python
print(f"Akurasi Model KNN : {skor_knn:.4f}")
print(f"Akurasi Model SVM : {skor_svm:.4f}")
```

```python
print("\nKESIMPULAN AKHIR ")

if skor_svm > skor_knn:
    print("=> Model Support Vector Machine (SVM) memberikan hasil akurasi yang lebih unggul dibandingkan dengan KNN pada dataset ini.")
elif skor_knn > skor_svm:
    print("=> Model K-Nearest Neighbors (KNN) memberikan hasil akurasi yang lebih unggul dibandingkan dengan SVM pada dataset ini.")
else:
    print("=> Kedua model, KNN dan SVM, memiliki performa akurasi yang persis sama pada dataset ini.")
```
